```python
# import necessary libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
# load data
titanic = pd.read_csv('Titanic_dataset.csv')
titanic.head()
```

```python
# drop unnecessary columns
df = titanic.drop(['PassengerId','Name','Ticket','Cabin'],axis = 1)
df
```

```python
# checking missing values
df.isna()
```

```python
# counting missing values
df.isna().any()
```

```python
# droping only the columns having missing values
df = df.dropna(subset=['Age','Embarked'])
df
```

```python
# re-checking for missing values
df.isna().any()
```

```python
# label encoding columns
df.loc[:,'Sex'] = LabelEncoder().fit_transform(df['Sex'])
df.loc[:,'Embarked'] = LabelEncoder().fit_transform(df['Embarked'])
```

```python
df
```

```python
# predictor and response variables
x = df.drop('Survived',axis=1)
y = df.Survived
```

```python
# splitting the data
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# model building
model = LogisticRegression(max_iter = 1000)
```

```python
# fitting the data into model
model.fit(x_train,y_train)
```

```python
# prediction
y_pred = model.predict(x_test)
```

```python
# resultant matrix
import numpy as np
result_matrix = np.column_stack((y_test, y_pred))
print('True Predicted')
print(result_matrix)
```

```python
accuracy = accuracy_score(y_test,y_pred)
print("Accuracy:",accuracy*100)
```

```python
# plotting the results
plt.scatter(y_test, y_pred, color='blue', label='Actual vs Predicted')
```
